## 10. Inspect Duplicate Records

```python
# Show duplicated records
df[df.duplicated(keep=False)].sort_values(
    by=["location", "year", "category", "rank", "query"]
).head(20)
```

### Interpretation — Duplicate Inspection

The duplicate rows are displayed above for manual inspection.

Because the duplicate test compares all columns, these records have identical values across `location`, `year`, `category`, `rank`, and `query`. If each occurrence represents the same real-world observation, keeping both copies could bias counts and frequency-based analysis.

A safe workflow is to **review first and remove confirmed accidental duplicates** rather than deleting them automatically during initial analysis.
